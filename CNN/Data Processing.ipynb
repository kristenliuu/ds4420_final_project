```python
import os
import numpy as np
from PIL import Image
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data_dir = 'Data_200_Imgs'

img_size = (224, 224)

images = []
labels = []

label_encoder = LabelEncoder()
all_labels = []

for label in os.listdir(data_dir):
    print(label)
    class_dir = os.path.join(data_dir, label)
    
    if os.path.isdir(class_dir):
        all_labels.append(label)

        for img_name in os.listdir(class_dir):
            img_path = os.path.join(class_dir, img_name)
            
            if img_path.endswith('.jpg') or img_path.endswith('.png'):
                img = Image.open(img_path).resize(img_size)
                img_array = np.array(img)
                
                images.append(img_array)
                labels.append(label)

label_encoder.fit(all_labels)
images = np.array(images)
encoded_labels = label_encoder.transform(labels).astype(int)

np.save('npy_and_labels_200/images.npy', images)  
np.savetxt('npy_and_labels_200/labels.csv', encoded_labels, delimiter=',')  
label_mapping = pd.DataFrame({'class': label_encoder.classes_, 'label': range(len(label_encoder.classes_))})
label_mapping.to_csv('npy_and_labels_200/label_mapping.csv', index=False)

print("Images and labels have been saved to 'images.npy' and 'labels.csv'")
```
